# Mini-projet : Classification de battements cardiaques ECG

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
```

## 1. Chargement du dataset ECG

```
# Exemple : si vous avez un CSV avec signaux et labels
# df = pd.read_csv('ecg_beats.csv')
# X = df.drop('label', axis=1).values
# y = df['label'].values

# Placeholder
X = np.random.randn(1000, 50)
y = np.random.randint(0, 3, 1000)
print(X.shape, y.shape)
```

## 2. Prétraitement (normalisation)

```
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 3. Découpage train/test

```
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=0, stratify=y)
```

## 4. Classificateur Bayésien Gaussien

```
clf_nb = GaussianNB()
clf_nb.fit(X_train, y_train)
y_pred_nb = clf_nb.predict(X_test)

print("=== Bayes ===")
print(confusion_matrix(y_test, y_pred_nb))
print(classification_report(y_test, y_pred_nb))
```

## 5. k-NN

```
clf_knn = KNeighborsClassifier(n_neighbors=5)
clf_knn.fit(X_train, y_train)
y_pred_knn = clf_knn.predict(X_test)

print("=== kNN ===")
print(confusion_matrix(y_test, y_pred_knn))
print(classification_report(y_test, y_pred_knn))
```
