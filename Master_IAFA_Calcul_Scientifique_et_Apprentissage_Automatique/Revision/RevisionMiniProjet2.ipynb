```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import math
from scipy import stats
import sklearn


# # Chargement des données


# Chargement des données d'apprentissage
Xtrain_Poignet=np.load('./Xtrain_Poignet_TP1.npy')
print('Taille de la base d apprentissage :', Xtrain_Poignet.shape)

# Chargement des données de test
Xtest_Poignet=np.load('./Xtest_Poignet_TP1.npy')

print('Taille de la base de test :', Xtest_Poignet.shape)

# CAPTEUR CHEVILLE
Xtrain_Cheville=np.load('./Xtrain_Cheville_TP1.npy')
print('Taille de la base d apprentissage :', Xtrain_Cheville.shape)

# Chargement des données de test
Xtest_Cheville=np.load('./Xtest_Cheville_TP1.npy')
print('Taille de la base de test :', Xtest_Cheville.shape)

# LABELS :
y_train=np.load('./LabelTrain_TP1.npy')
y_test=np.load('./LabelTest_TP1.npy')
print('Taille label apprentissage :',y_train.shape, 'Taille label test :',y_test.shape)
```

# # Partie I : Classification des activités sans prétraitement
#
# Réaliser une classification des activités par deux méthodes d'apprentissage supervisé : les kppv du TP2 et la classification bayésienne du TP4 avec la fonction de scikit-learn GaussianNB.
#
#
# ## Méthodes supervisées  :
# Indiquer  les paramètres utilisés pour les méthodes de classification.  Calculer la matrice de confusion associée, le pourcentage de bonne classification et commentez les résultats.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

# Paramètres choisis (simples et standards pour un TP)
k = 5  # k pour k-NN

# Instanciation des classifieurs
knn = KNeighborsClassifier(n_neighbors=k)
nb  = GaussianNB()
```

# ### Premier capteur :


```python
print("\n--- Capteur : POIGNET (sans prétraitement) ---")

# k-NN sur données brutes (poignet)
knn.fit(Xtrain_Poignet, y_train)
y_pred_knn_poignet = knn.predict(Xtest_Poignet)
cm_knn_poignet = confusion_matrix(y_test, y_pred_knn_poignet)
acc_knn_poignet = accuracy_score(y_test, y_pred_knn_poignet)

print("k-NN (k={}) - Matrice de confusion (Poignet):".format(k))
print(cm_knn_poignet)
print("k-NN (Poignet) - Pourcentage de bonne classification: {:.2f}%".format(100*acc_knn_poignet))
print(classification_report(y_test, y_pred_knn_poignet, digits=4))

# GaussianNB sur données brutes (poignet)
nb.fit(Xtrain_Poignet, y_train)
y_pred_nb_poignet = nb.predict(Xtest_Poignet)
cm_nb_poignet = confusion_matrix(y_test, y_pred_nb_poignet)
acc_nb_poignet = accuracy_score(y_test, y_pred_nb_poignet)

print("GaussianNB - Matrice de confusion (Poignet):")
print(cm_nb_poignet)
print("GaussianNB (Poignet) - Pourcentage de bonne classification: {:.2f}%".format(100*acc_nb_poignet))
print(classification_report(y_test, y_pred_nb_poignet, digits=4))
```


# ## Réduction de dimension par Analyse en Composantes Principales (ACP)
#
# Réaliser une ACP pour réduire la dimension et visualiser les données. Discuter sur le choix de la dimension.



```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

print("\n\n--- Partie II : Prétraitement (StandardScaler + PCA) ---")

# Standardisation avant PCA (très important)
scaler_poignet = StandardScaler()
Xtrain_Poignet_scaled = scaler_poignet.fit_transform(Xtrain_Poignet)
Xtest_Poignet_scaled = scaler_poignet.transform(Xtest_Poignet)

scaler_cheville = StandardScaler()
Xtrain_Cheville_scaled = scaler_cheville.fit_transform(Xtrain_Cheville)
Xtest_Cheville_scaled = scaler_cheville.transform(Xtest_Cheville)

# Déterminer le nombre de composantes pour expliquer 95% de la variance
pca_temp = PCA()
pca_temp.fit(Xtrain_Poignet_scaled)
cumvar_poignet = np.cumsum(pca_temp.explained_variance_ratio_)
n_comp_poignet_95 = np.searchsorted(cumvar_poignet, 0.95) + 1
# Pour rester raisonnable en TP, si c'est trop grand limiter à 50
n_comp_poignet = min(n_comp_poignet_95, 50)

pca_temp.fit(Xtrain_Cheville_scaled)
cumvar_cheville = np.cumsum(pca_temp.explained_variance_ratio_)
n_comp_cheville_95 = np.searchsorted(cumvar_cheville, 0.95) + 1
n_comp_cheville = min(n_comp_cheville_95, 50)

print("Poignet : composantes pour 95% variance (avant cap):", n_comp_poignet_95)
print("Poignet : choix retenu (cap 50):", n_comp_poignet)
print("Cheville : composantes pour 95% variance (avant cap):", n_comp_cheville_95)
print("Cheville : choix retenu (cap 50):", n_comp_cheville)
```

# ## Premier capteur :  méthodes supervisées avec réduction de dimension par ACP sur l'ensemble d'apprentissage
# Après avoir réalisé une ACP sur l'ensemble d'apprentissage pour réduire la dimension puis projeter les données de l'ensemble de test dans la base de l'ACP, appliquez les deux méthodes supervisées.
#
#

```python
print("\n--- POIGNET : PCA + classification ---")
pca_poignet = PCA(n_components=n_comp_poignet)
Xtrain_Poignet_pca = pca_poignet.fit_transform(Xtrain_Poignet_scaled)
Xtest_Poignet_pca = pca_poignet.transform(Xtest_Poignet_scaled)

print("Shape après PCA (Poignet) - train:", Xtrain_Poignet_pca.shape, " test:", Xtest_Poignet_pca.shape)

# Classifieurs simples sur données réduites
knn_pca = KNeighborsClassifier(n_neighbors=k)
nb_pca  = GaussianNB()

# k-NN
knn_pca.fit(Xtrain_Poignet_pca, y_train)
y_pred_knn_poignet_pca = knn_pca.predict(Xtest_Poignet_pca)
cm_knn_poignet_pca = confusion_matrix(y_test, y_pred_knn_poignet_pca)
acc_knn_poignet_pca = accuracy_score(y_test, y_pred_knn_poignet_pca)

print("k-NN (Poignet + PCA) - Matrice de confusion:")
print(cm_knn_poignet_pca)
print("k-NN (Poignet + PCA) - Pourcentage de bonne classification: {:.2f}%".format(100*acc_knn_poignet_pca))
print(classification_report(y_test, y_pred_knn_poignet_pca, digits=4))

# GaussianNB
nb_pca.fit(Xtrain_Poignet_pca, y_train)
y_pred_nb_poignet_pca = nb_pca.predict(Xtest_Poignet_pca)
cm_nb_poignet_pca = confusion_matrix(y_test, y_pred_nb_poignet_pca)
acc_nb_poignet_pca = accuracy_score(y_test, y_pred_nb_poignet_pca)

print("GaussianNB (Poignet + PCA) - Matrice de confusion:")
print(cm_nb_poignet_pca)
print("GaussianNB (Poignet + PCA) - Pourcentage de bonne classification: {:.2f}%".format(100*acc_nb_poignet_pca))
print(classification_report(y_test, y_pred_nb_poignet_pca, digits=4))
```

En Analyse en Composantes Principales (PCA), la variance mesure la quantité d’information conservée dans les nouvelles composantes.

Chaque composante principale capture une partie de la variabilité totale des signaux.

Plus une composante a une grande variance, plus elle transporte une information utile pour différencier les classes.

À l’inverse, les composantes avec très faible variance capturent principalement du bruit, des détails inutiles, ou des redondances.

➡️ Quand on dit que 144 composantes sont nécessaires pour conserver 95% de la variance au poignet, cela signifie :

Les données du poignet sont très dispersées dans l’espace,

Il existe beaucoup de dimensions qui contiennent un peu d’information mais aussi du bruit,

Les signaux sont moins structurés → classification plus difficile.

➡️ Quand la cheville demande 192 composantes pour 95% de variance :

L’activité mesurée a encore plus de variabilité utile,

Les mouvements captés sont plus riches → d’où les meilleures performances pour k-NN.

⚠️ Le fait de choisir 50 composantes (au lieu de 144 ou 192) revient à :

Garder uniquement les composantes les plus informatives,

Retirer beaucoup de bruit,

Rendre les distances euclidiennes plus fiables pour les méthodes comme k-NN.

🎯 Au final :

La variance en PCA indique combien de dimensions sont réellement utiles.

Si beaucoup de variance est concentrée dans peu de composantes → données bien structurées.

Si la variance est étalée sur beaucoup de composantes → données plus bruyantes ou complexes.
