Représenter l'histogramme associé $k=$ nombre de face pour six lancés de pièces consécutifs

```python
k_values
probabilities
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import binom

# Paramètres
n = 6  # Nombre de lancers
p = 0.5  # Probabilité d'obtenir une face (F) lors d'un lancer

# Créer une distribution binomiale
distribution = binom(n, p)

# Calculer les probabilités pour chaque nombre de faces possibles
k_values = np.arange(0, n+1)
probabilities = distribution.pmf(k_values)

# Créer l'histogramme
plt.bar(k_values, probabilities, align='center')
plt.xlabel('Nombre de faces (F)')
plt.ylabel('Probabilité')
plt.title('Histogramme du nombre de faces obtenues lors de six lancés de pièces')
plt.xticks(k_values)
plt.show()
```

Approximation de la loi binomiale en TCL (théorème de Moivre Laplace)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Binomial distribution parameters
n = 50        # Number of trials
p = 0.5       # Probability of success

# Calculate mean (μ) and standard deviation (σ) of the binomial distribution
mean_binomial = n * p
std_binomial = np.sqrt(n * p * (1 - p))

# Generate random samples from the binomial distribution
binomial_samples = np.random.binomial(n, p, 1000)

# Standardize the samples to get the standardized normal distribution
standardized_samples = (binomial_samples - mean_binomial) / std_binomial

# Plot the histograms of the binomial distribution and the standardized normal distribution
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.hist(binomial_samples, bins=20, color='blue', alpha=0.6)
plt.title('Binomial Distribution (n=50, p=0.5)')
plt.xlabel('Number of Successes')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(standardized_samples, bins=20, density=True, color='green', alpha=0.6)
plt.title('Standardized Normal Distribution')
plt.xlabel('Standardized Value (Z)')
plt.ylabel('Density')

# Plot the density of the standardized normal distribution using scipy.stats.norm
x_range = np.linspace(-4, 4, 1000)
standard_normal_density = norm.pdf(x_range, 0, 1)
plt.plot(x_range, standard_normal_density, color='red', lw=2, label='Standard Normal Density')
plt.legend()

plt.tight_layout()
plt.show()
```

Affichage de deux courbre de densité de loi normal aux paramètres différents

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters for the first normal distribution
mu1 = 0        # Mean
sigma1 = 1     # Standard deviation

# Parameters for the second normal distribution
mu2 = 2        # Mean
sigma2 = 0.5   # Standard deviation

# Generate random samples for both distributions
num_samples = 1000
samples1 = np.random.normal(mu1, sigma1, num_samples)
samples2 = np.random.normal(mu2, sigma2, num_samples)

# Calculate the density values for both distributions
x_range = np.linspace(-5, 5, 1000)
density1 = norm.pdf(x_range, mu1, sigma1)
density2 = norm.pdf(x_range, mu2, sigma2)

# Plot the first normal distribution
plt.plot(x_range, density1, label=f'N(μ={mu1}, σ={sigma1})', color='blue')
plt.hist(samples1, bins=30, density=True, alpha=0.6, color='blue')

# Plot the second normal distribution
plt.plot(x_range, density2, label=f'N(μ={mu2}, σ={sigma2})', color='green')
plt.hist(samples2, bins=30, density=True, alpha=0.6, color='green')

# Add legend and labels
plt.legend()
plt.title('Two Normal Distributions')
plt.xlabel('X')
plt.ylabel('Density')

plt.tight_layout()
plt.show()
```

Comparaison de $\mathbb{P}[|X-\mu|<\sigma$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters for the first normal distribution
mu1 = 0        # Mean
sigma1 = 1     # Standard deviation

# Parameters for the second normal distribution
mu2 = 2        # Mean
sigma2 = 0.5   # Standard deviation

# Generate random samples for both distributions
num_samples = 1000
samples1 = np.random.normal(mu1, sigma1, num_samples)
samples2 = np.random.normal(mu2, sigma2, num_samples)

# Calculate the density values for both distributions
x_range = np.linspace(-5, 5, 1000)
density1 = norm.pdf(x_range, mu1, sigma1)
density2 = norm.pdf(x_range, mu2, sigma2)

# Plot the first normal distribution
plt.plot(x_range, density1, label=f'N(μ={mu1}, σ={sigma1})', color='blue')
plt.hist(samples1, bins=30, density=True, alpha=0.6, color='blue')

# Calculate the absolute difference between the samples and the mean for the first distribution
abs_diff1 = np.abs(samples1 - mu1)

# Calculate the probability P[|X - μ| < σ] for the first distribution
prob1 = np.sum(abs_diff1 < sigma1) / len(samples1)

# Add hatching for the region representing P[|X - μ| < σ] for the first distribution
plt.axvspan(mu1 - sigma1, mu1 + sigma1, facecolor='blue', alpha=0.3, hatch='/', label=f'P[|X - μ| < σ] = {prob1:.2f}')

# Plot the second normal distribution
plt.plot(x_range, density2, label=f'N(μ={mu2}, σ={sigma2})', color='green')
plt.hist(samples2, bins=30, density=True, alpha=0.6, color='green')

# Calculate the absolute difference between the samples and the mean for the second distribution
abs_diff2 = np.abs(samples2 - mu2)

# Calculate the probability P[|X - μ| < σ] for the second distribution
prob2 = np.sum(abs_diff2 < sigma2) / len(samples2)

# Add hatching for the region representing P[|X - μ| < σ] for the second distribution
plt.axvspan(mu2 - sigma2, mu2 + sigma2, facecolor='green', alpha=0.3, hatch='\\', label=f'P[|X - μ| < σ] = {prob2:.2f}')

# Add legend and labels
plt.legend()
plt.title('Two Normal Distributions with Probability P[|X - μ| < σ]')
plt.xlabel('X')
plt.ylabel('Density')

plt.tight_layout()
plt.show()
```

plot z area

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Range of z-values (standard deviations from the mean)
z_range = np.linspace(-3, 3, 1000)

# Calculate the probability density function (PDF) of the standard normal distribution
pdf_values = norm.pdf(z_range, loc=0, scale=1)

# Plot the density function of the centered standard normal distribution
plt.plot(z_range, pdf_values, color='blue', label='PDF')

# Define the value of 'a'
a = 1.5

# Calculate the cumulative distribution function (CDF) up to 'a'
cdf_a = norm.cdf(a, loc=0, scale=1)

# Fill the area under the PDF curve for P[Z < a]
plt.fill_between(z_range, pdf_values, where=z_range <= a, color='gray', alpha=0.5, label=f'P[Z < {a}] = {cdf_a:.2f}')

# Add labels and legend
plt.title(f'P[Z < {a}] on Density Function of Centered Standard Normal Distribution')
plt.xlabel('Z (Standard Deviations from the Mean)')
plt.ylabel('Probability Density')
plt.legend(loc='upper left')

plt.grid()
plt.tight_layout()
plt.show()
```
