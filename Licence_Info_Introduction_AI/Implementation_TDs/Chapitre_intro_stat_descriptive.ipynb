Seaborn propose plusieurs jeux de données intégrés (DataFrames) qui peuvent être utilisés pour pratiquer et illustrer des concepts d'analyse de données et de visualisation. Voici quelques-uns des jeux de données couramment utilisés dans Seaborn :

    iris: Un jeu de données classique contenant des mesures de longueur et de largeur de pétales et de sépales pour trois espèces différentes d'iris.

    titanic: Un jeu de données sur les passagers du Titanic, comprenant des informations telles que la classe de voyage, le sexe, l'âge, etc.

    tips: Un jeu de données sur les pourboires dans les restaurants, comprenant des informations sur le montant du repas, le pourboire, le sexe du payeur, etc.

    flights: Un jeu de données contenant des informations mensuelles sur les vols.

    planets: Un jeu de données sur les exoplanètes découvertes, comprenant des informations sur la masse, la période orbitale, etc.

    diamonds: Un jeu de données sur les caractéristiques des diamants, y compris le carat, la coupe, la clarté, la couleur, etc.

    exercise: Un jeu de données contenant des informations sur différentes activités d'exercice et leurs effets sur la fréquence cardiaque, le temps d'exercice, etc.

    seaborn_data: Un jeu de données synthétiques généré par Seaborn lui-même pour des exemples et des démonstrations.

tableau d'effectif

```python
import pandas as pd
from tabulate import tabulate

# Exemple de données
donnees = [5, 7, 8, 5, 7, 5, 6, 8, 7, 5, 8, 7, 6, 5, 8]

# Création d'un DataFrame avec les données
df = pd.DataFrame({'Valeurs': donnees})

# Calcul des effectifs
effectifs = df['Valeurs'].value_counts().sort_index()

# Création d'une liste de tuples pour le tabulateur
tableau_effectifs = [(valeur, effectif) for valeur, effectif in zip(effectifs.index, effectifs)]

# Personnalisation des options de tabulate
options = {
    'tablefmt': 'fancy_grid',
    'headers': ['Valeur', 'Effectif'],
    'numalign': 'center',
    'stralign': 'center',
    'showindex': False
}

# Génération et affichage du tableau avec tabulate
tableau_formatte = tabulate(tableau_effectifs, **options)
print(tableau_formatte)

```

Si vous avez besoin de plus de personnalisation ou d'autres formats de tableau, consultez la documentation de la bibliothèque tabulate pour voir les options disponibles : https://pypi.org/project/tabulate/

```python
# Calcul des effectifs
effectifs = df['Valeurs'].value_counts().sort_index()
total = len(donnees)
effectifs['Fréquence'] = effectifs / total

# Création d'une liste de tuples pour le tabulateur
tableau_effectifs = [(valeur, effectif, f'{frequence:.2%}') for valeur, effectif, frequence in zip(effectifs.index, effectifs, effectifs['Fréquence'])]

# Personnalisation des options de tabulate
options = {
    'tablefmt': 'fancy_grid',
    'headers': ['Valeur', 'Effectif', 'Fréquence'],
    'numalign': 'center',
    'stralign': 'center',
    'showindex': False
}

# Génération et affichage du tableau avec tabulate
tableau_formatte = tabulate(tableau_effectifs, **options)
print(tableau_formatte)
```

Création de classe de valeurs

```python
import pandas as pd
import matplotlib.pyplot as plt

# Exemple de données de notes d'étudiants
notes = [75, 82, 90, 68, 78, 88, 72, 95, 85, 92, 78, 80, 88, 92, 76, 81, 87, 84, 79, 90]

# Création d'un DataFrame avec les données
df = pd.DataFrame({'Notes': notes})

# Définition des intervalles de classes pour l'histogramme
classes = [60, 70, 80, 90, 100]

# Création d'une colonne de classes pour les données
df['Classe'] = pd.cut(df['Notes'], bins=classes, right=False)

# Calcul des effectifs pour chaque classe
effectifs = df['Classe'].value_counts().sort_index()

# Affichage des effectifs
print("Tableau d'effectifs par classe :")
print(effectifs)

# Affichage de l'histogramme
plt.hist(df['Notes'], bins=classes, edgecolor='black', alpha=0.7)
plt.xlabel('Notes')
plt.ylabel('Effectif')
plt.title('Histogramme des notes d\'étudiants')
plt.grid(True)
plt.show()
```

Histogramme de proportion/fréquence

```python
import pandas as pd
import matplotlib.pyplot as plt

# Exemple de données de notes d'étudiants
notes = [75, 82, 90, 68, 78, 88, 72, 95, 85, 92, 78, 80, 88, 92, 76, 81, 87, 84, 79, 90]

# Création d'un DataFrame avec les données
df = pd.DataFrame({'Notes': notes})

# Définition des intervalles de classes pour l'histogramme
classes = [60, 70, 80, 90, 100]

# Création d'une colonne de classes pour les données
df['Classe'] = pd.cut(df['Notes'], bins=classes, right=False)

# Calcul des proportions pour chaque classe
proportions = (df['Classe'].value_counts() / len(df)).sort_index()

# Affichage des proportions
print("Tableau des proportions par classe :")
print(proportions)

# Obtention des positions de départ pour les barres
positions = range(len(proportions))

# Affichage de l'histogramme de proportions avec les rectangles collés
plt.bar(positions, proportions.values, tick_label=proportions.index, edgecolor='black', alpha=0.7)
plt.xlabel('Classes de notes')
plt.ylabel('Proportion')
plt.title('Histogramme de proportions des notes d\'étudiants')
plt.grid(True)
plt.show()
```

Meusures de dispersion

 Dessiner et comparer les deux boxplot associée à ces notes deux cours :
C1 = \{04,12,23,35,46,52,67,75,83,92\} et C2 = \{18,19,21,23,24,25,27,29,30,93\}}.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Données des deux cours
C1 = [4, 12, 23, 35, 46, 52, 67, 75, 83, 92]
C2 = [18, 19, 21, 23, 24, 25, 27, 29, 30, 93]

# Créer un DataFrame pour les données
data = {'Cours': ['C1'] * len(C1) + ['C2'] * len(C2),
        'Notes': C1 + C2}
df = pd.DataFrame(data)

# Créer le boxplot
plt.figure(figsize=(8, 6))
sns.boxplot(x='Cours', y='Notes', data=df)
plt.title("Comparaison des boxplots pour les deux cours")
plt.show()
```
