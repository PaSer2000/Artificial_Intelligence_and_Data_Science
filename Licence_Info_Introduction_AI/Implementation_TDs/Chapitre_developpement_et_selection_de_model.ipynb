```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from sklearn.metrics import r2_score

# Générer des données synthétiques pour l'exemple
np.random.seed(0)
X = np.sort(5 * np.random.rand(100, 1), axis=0)
y = np.sin(X).ravel() + np.random.normal(0, 0.1, X.shape[0])

# Diviser les données en ensembles d'entraînement et de test
X_train, X_test = X[:60], X[60:]
y_train, y_test = y[:60], y[60:]

# Modèle de régression linéaire (Underfitting)
def linear_model(x, a, b):
    return a * x + b

# Modèle de régression polynomiale d'ordre 2 (Bon ajustement)
def quadratic_model(x, a, b, c):
    return a * x**2 + b * x + c

# Modèle sinusoidal (Bon ajustement)
def sinusoidal_model(x, a, b, c, d):
    return a * np.sin(b * x + c) + d

# Ajuster le modèle linéaire aux données d'entraînement
params_linear, _ = curve_fit(linear_model, X_train.ravel(), y_train)
a_linear, b_linear = params_linear
y_pred_linear_train = linear_model(X_train.ravel(), a_linear, b_linear)
y_pred_linear_test = linear_model(X_test.ravel(), a_linear, b_linear)
r2_linear_train = r2_score(y_train, y_pred_linear_train)

r2_linear_test = r2_score(y_test, y_pred_linear_test)

# Ajuster le modèle polynomial aux données d'entraînement
params_quadratic, _ = curve_fit(quadratic_model, X_train.ravel(), y_train)
a_quadratic, b_quadratic, c_quadratic = params_quadratic
y_pred_quadratic_train = quadratic_model(X_train.ravel(), a_quadratic, b_quadratic, c_quadratic)
y_pred_quadratic_test = quadratic_model(X_test.ravel(), a_quadratic, b_quadratic, c_quadratic)
r2_quadratic_train = r2_score(y_train, y_pred_quadratic_train)
r2_quadratic_test = r2_score(y_test, y_pred_quadratic_test)

# Ajuster le modèle sinusoidal aux données d'entraînement
params_sinusoidal, _ = curve_fit(sinusoidal_model, X_train.ravel(), y_train)
a_sinusoidal, b_sinusoidal, c_sinusoidal, d_sinusoidal = params_sinusoidal
y_pred_sinusoidal_train = sinusoidal_model(X_train.ravel(), a_sinusoidal, b_sinusoidal, c_sinusoidal, d_sinusoidal)
y_pred_sinusoidal_test = sinusoidal_model(X_test.ravel(), a_sinusoidal, b_sinusoidal, c_sinusoidal, d_sinusoidal)
r2_sinusoidal_train = r2_score(y_train, y_pred_sinusoidal_train)
r2_sinusoidal_test = r2_score(y_test, y_pred_sinusoidal_test)

# Affichage des résultats
print("R² en Train (Underfitting):", r2_linear_train)
print("R² en Test (Underfitting):", r2_linear_test)

print("R² en Train (Bon ajustement - Sinusoïdal):", r2_sinusoidal_train)
print("R² en Test (Bon ajustement - Sinusoïdal):", r2_sinusoidal_test)

print("R² en Train (Overfitting - Polynomiale):", r2_quadratic_train)
print("R² en Test (Overfitting - Polynomiale):", r2_quadratic_test)
```

Lorsque le R² est négatif, cela signifie que le modèle est pire que la ligne horizontale moyenne. En d'autres termes, il ne parvient pas du tout à capturer la tendance des données. C'est un exemple typique d'un modèle qui a appris le bruit dans les données d'entraînement plutôt que la relation sous-jacente.

```python
import numpy as np
x = np.arange(50, dtype=float)
y = x
y_bar = np.mean(y)
def m1(x): return 0
def m2(x): return y_bar

rss1 = np.sum((y - m1(x)) ** 2)
tss = np.sum((y - y.mean()) ** 2)
print(1 - rss1 / tss)
rss2 = np.sum((y - m2(x)) ** 2)
print(1 - rss2 / tss)

```

Bagging (Bootstrap Aggregating) avec un modèle de régression

Le bagging consiste à entraîner plusieurs modèles indépendants sur des sous-ensembles aléatoires de données et à agréger leurs prédictions pour obtenir une prédiction finale plus robuste.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import BaggingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

# Générer des données synthétiques pour l'exemple
X, y = make_regression(n_samples=100, n_features=1, noise=0.1, random_state=0)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Créer un modèle de régression de base (arbre de décision)
base_model = DecisionTreeRegressor()

# Créer un modèle de Bagging avec 10 réplicats
bagging_model = BaggingRegressor(base_model, n_estimators=10, random_state=0)

# Entraîner le modèle Bagging sur les données d'entraînement
bagging_model.fit(X_train, y_train)

# Prédire avec le modèle Bagging
y_pred = bagging_model.predict(X_test)

# Calculer et afficher la MSE
mse = mean_squared_error(y_test, y_pred)
print(f"MSE du modèle Bagging : {mse:.2f}")

# Afficher les prédictions et les données réelles
plt.scatter(X_test, y_test, label='Données réelles', alpha=0.7)
plt.scatter(X_test, y_pred, label='Prédictions', alpha=0.7)
plt.legend()
plt.xlabel('X')
plt.ylabel('y')
plt.title('Bagging Regressor')
plt.show()
```

Boosting avec un modèle de régression

Le boosting consiste à combiner plusieurs modèles faibles pour former un modèle fort.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import AdaBoostRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

# Générer des données synthétiques pour l'exemple
X, y = make_regression(n_samples=100, n_features=1, noise=0.1, random_state=0)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Créer un modèle de régression de base (arbre de décision)
base_model = DecisionTreeRegressor(max_depth=2)

# Créer un modèle de Boosting (AdaBoost) avec 100 réplicats
boosting_model = AdaBoostRegressor(base_model, n_estimators=100, random_state=0)

# Entraîner le modèle de Boosting sur les données d'entraînement
boosting_model.fit(X_train, y_train)

# Prédire avec le modèle de Boosting
y_pred = boosting_model.predict(X_test)

# Calculer et afficher la MSE
mse = mean_squared_error(y_test, y_pred)
print(f"MSE du modèle de Boosting : {mse:.2f}")

# Afficher les prédictions et les données réelles
plt.scatter(X_test, y_test, label='Données réelles', alpha=0.7)
plt.scatter(X_test, y_pred, label='Prédictions', alpha=0.7)
plt.legend()
plt.xlabel('X')
plt.ylabel('y')
plt.title('AdaBoost Regressor')
plt.show()
```
