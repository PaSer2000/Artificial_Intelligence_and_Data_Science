```python
import numpy as np
import matplotlib.pyplot as plt

# Set the parameters
p = 0.5
num_points = 100

# Generate 100 data points from Bernoulli distribution
data_points = np.random.binomial(1, p, num_points)

# Calculate the sample mean for each subset of the data
sample_means = np.cumsum(data_points) / np.arange(1, num_points + 1)

# Calculate the standard error of the mean (SEM)
sem = np.sqrt(sample_means * (1 - sample_means) / np.arange(1, num_points + 1))

# Calculate 95% confidence intervals
z_score = 1.96  # 95% confidence level corresponds to a z-score of 1.96
lower_ci = sample_means - z_score * sem
upper_ci = sample_means + z_score * sem

# Plotting the results
plt.figure(figsize=(10, 6))
plt.errorbar(np.arange(1, num_points + 1), sample_means, yerr=z_score * sem, fmt='o', markersize=3, label='Sample Mean with 95% CI')
plt.axhline(p, color='red', linestyle='dashed', label='True Parameter (p=0.5)')
plt.xlabel('Number of Data Points')
plt.ylabel('Sample Mean')
plt.legend()
plt.title('Sample Means and 95% Confidence Intervals')
plt.show()
```

t de student avec différent n et n-1 degré de liberté à 95%

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t

# Parameters for the t-student distributions
dof_values = [3, 10, 30]  # Degrees of freedom (n-1) for the distributions

# Generate x-axis values for the plot
x = np.linspace(-5, 5, 1000)

# Plot each t-student distribution
for dof in dof_values:
    t_dist = t(df=dof)  # Create a t-student distribution object with given dof
    plt.plot(x, t_dist.pdf(x), label=f't-distribution (n={dof+1})')

# Add labels and legend
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.title('T-Student Distributions with n-1 Degrees of Freedom')
plt.legend()

# Show the plot
plt.show()
```
