Simulation d'une variable de Bernouilli

```python
x = np.random.choice(np.arange(0,2),10000, p=[0.5,0.5])
print(x)
plt.hist(x, bins=[0,0.9,1,1.9])
plt.show()
```

Simulation d'une Binomiale

```python
#Liste vide ou l'on va sauvgarder la somme des observations réalisées
experiments = []
xp = 5

while xp>0:
  experiments.append(np.sum(np.random.choice(np.arange(0, 2), 10, p=[0.5, 0.5])))
  xp -= 1

print(experiments)
plt.hist(experiments, bins=np.arange(11))
plt.show()
```

Approximation de la Binomiale par la Normal

```python
# On réalise un nombre d'expérimentations importantes
experiments = []
xp = 100

while xp>0:
  experiments.append(np.sum(np.random.choice(np.arange(0, 2), 10, p=[0.5, 0.5])))
  xp -= 1

print(experiments)
plt.hist(experiments, bins=np.arange(11))
plt.show()
```

Sommation de variable aléatoire

Représenter graphiquement la fonction $f_1$ associée à la densité de probabilité de la variable $X_1$ issue d'une loi uniforme sur [0,1]. Générer également la fonction $f_2$ associée à la variable $X_2$ issue d'une loi uniforme sur [0,1]. Représenter la fonction $f_3$ tel que $X_3=X_1+X_2$

```python
from scipy.stats import uniform
import matplotlib.pyplot as plt
import numpy as np

x1 = uniform.rvs(0.01,0.99,size=1000)

plt.hist(x1,density = True)
plt.axhline(y=uniform.pdf(x1[0]),color='r')
plt.title('Uniform distribution')
plt.ylabel('Density')
plt.xlabel('X')
plt.show()

```

```python
# Import the libraries
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(x1, kde=True)
```

```python
x2 = np.random.uniform(0, 1, 1000)
x3 = x1 + x2

sns.histplot(x3, kde=True)
```
